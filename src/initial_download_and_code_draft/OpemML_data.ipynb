```python
def print_unique_counts(df):
    data_vals = {}
    for col in df.columns:
        try:
            unique_count = df[col].astype(str).nunique(dropna=False)
            print(f"{col}: {unique_count}")
            data_vals[col] = unique_count
        except Exception as e:
            print(f"{col}: ERROR ({e})")
    return data_vals
```

```python
def dataset_to_text(row):
    parts = [
        f"Dataset: {row['name']}",
        f"Description: {row['description']}",
        f"Tags: {', '.join(row['tag']) if isinstance(row['tag'], list) else row['tag']}",
        f"Target: {row['default_target_attribute']}",
        f"Language: {row['language']}",
        f"Creator: {row['creator']}",
    ]
    return "\n".join(str(x) for x in parts if x and str(x) != "nan")
```

```python
df["embedding_text"] = df.apply(dataset_to_text, axis=1)
```

```python
def recommend(idx, pred_matrix=embeddings, top_k=10):

    scores = list(enumerate(pred_matrix[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    top = scores[1:top_k+1]

    indices = [i for i, _ in top]

    
    return pd.DataFrame({
        "idx": indices,
        "name": df.iloc[indices]["name"].values,
        "description": df.iloc[indices]["description"].values,
        "tag": df.iloc[indices]["tag"].values,
        
    })
```
